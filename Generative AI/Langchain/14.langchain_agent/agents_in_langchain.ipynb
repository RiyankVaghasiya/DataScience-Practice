```python
from langchain_groq import ChatGroq
from langchain_community.tools import tool
import requests
from langchain_community.tools import DuckDuckGoSearchRun
import json
```

```python
search_tool = DuckDuckGoSearchRun()
```

```python
@tool
def get_weather_data(city: str) -> str:
  """this function fetches the current weather data of a given city"""

  url = f'https://api.weatherstack.com/current?access_key=c57cf965b1ac3377cb0798ae8e57e971&query={city}'

  response = requests.get(url)
  return response.json
```

```python
llm = ChatGroq(model='llama3-8b-8192')
```

```python
from langchain.agents import create_react_agent, AgentExecutor
from langchain import hub
```

```python
prompt = hub.pull("hwchase17/react")
```

```python
#creating the react agent manually with pulled prompt
agent = create_react_agent(
    llm = llm,
    tools= [search_tool,get_weather_data],
    prompt=prompt
)
```

```python
# wrap it with AgentExecutor
agent_executor = AgentExecutor(
    agent = agent,
    tools= [search_tool,get_weather_data],
    verbose= True
)
```

```python
#invoke
response = agent_executor.invoke({"input": "Find the capital of india and find it's current weather"})
print(response)
```
