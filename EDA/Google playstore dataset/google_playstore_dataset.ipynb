```python
df_copy = df_copy.drop(df_copy.index[10472])
```

```python
df_copy['Reviews'] = df_copy['Reviews'].astype(int)
```

```python
df_copy['Size'].unique()
```

```python
#now we will convert million to thousand
df_copy['Size'] = df_copy['Size'].str.replace('M',"000")
df_copy['Size'] = df_copy['Size'].str.replace('k','')
df_copy['Size'] = df_copy['Size'].replace('Varies with device',np.nan)
df_copy['Size'] = df_copy['Size'].astype(float)
```

```python
df_copy['Size'].unique()
```

```python
df_copy.info()
```

```python
df_copy.head(2)
```

```python
df_copy['Installs'].unique()
```

```python
df_copy['Price'].unique()
```

```python
df_copy['Installs'] = df_copy['Installs'].str.replace('+','')
df_copy['Price'] = df_copy['Price'].str.replace('$','')
df_copy['Installs'] = df_copy['Installs'].str.replace(',','')
```

```python
df_copy['Price'].unique()
```

```python
df_copy['Installs'].unique()
```

```python
df_copy['Price'] = df_copy['Price'].astype(float)
df_copy['Installs'] = df_copy['Installs'].astype(int)
```

```python
df_copy.info()
```

```python
df_copy.head(2)
```

```python
df_copy['Last Updated'] = pd.to_datetime(df_copy['Last Updated'])
```

```python
df_copy['last_updated_day'] = df_copy['Last Updated'].dt.day
df_copy['last_updated_month'] = df_copy['Last Updated'].dt.month
df_copy['last_updated_year'] = df_copy['Last Updated'].dt.year
```

```python
df_copy.drop(columns='Last Updated',index = 1,inplace = True)
```
